```python
def create_horizontal_task_plot(task_values, task_errors, task_annotations, task_title):
    models = list(task_values.keys())
    
    fig = plt.figure(figsize=(10, 8), dpi=150)
    
    gs = fig.add_gridspec(3, 1, height_ratios=[1, 1, 1], hspace=0.4)
    axs = [fig.add_subplot(gs[i]) for i in range(3)]
    
    fig.suptitle(task_title, fontweight='bold', fontsize=15, y=0.96)
    
    metrics_names = ["Accuracy", "F1-macro", "AUROC"]  # Define the names of metrics
    
    for i, metric in enumerate(key_metrics):
        ax = axs[i]
        
        metric_values = {model: task_values[model][i] for model in models}
        metric_errors = {model: task_errors[model][i] for model in models}
        metric_annotations = {model: task_annotations[model][i] for model in models}
        
        sorted_models = sorted(models, key=lambda m: metric_values[m], reverse=False)
        
        y_pos = np.arange(len(sorted_models))
        bars = ax.barh(y_pos, [metric_values[m] for m in sorted_models], 
                       color=[model_colors[m] for m in sorted_models],
                       edgecolor='black', linewidth=0.8, alpha=0.85, height=0.65)
        
        for j, model in enumerate(sorted_models):
            ax.errorbar(
                metric_values[model], j, 
                xerr=metric_errors[model], 
                fmt='none', ecolor='#888888', 
                capsize=3, capthick=1, linewidth=1
            )
        
        for j, bar in enumerate(bars):
            model = sorted_models[j]
            width = bar.get_width()
            annotation = metric_annotations[model]
            
            fontweight = 'bold' if model == "JINAv3F" else 'normal'
            
            ax.text(
                width + metric_errors[model] + 2, 
                bar.get_y() + bar.get_height()/2,
                f'{width:.1f}$^{{{annotation}}}$', ha='left', va='center',
                fontsize=13, fontweight=fontweight
            )
        
        ax.set_yticks(y_pos)
        ax.set_yticklabels(sorted_models)
        
        for j, model in enumerate(sorted_models):
            if model == "JINAv3F":
                ax.get_yticklabels()[j].set_weight('bold')
                ax.get_yticklabels()[j].set_color(model_colors["JINAv3F"])
        
        ax.set_ylabel(metrics_names[i], fontsize=13, fontweight='bold', labelpad=10)
        
        ax.grid(False)
        
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        ax.spines['left'].set_visible(False)
        
        max_value = max(metric_values[m] + metric_errors[m] for m in models)
        ax.set_xlim(0, max_value + 12)
    
    axs[-1].set_xlabel('Performance (%)', fontweight='bold', labelpad=8)
        
    effect_size_text = "Effect sizes vs. JINAv3: small (§), medium (†), large (‡), small increase (§). All metrics in %."
    fig.text(0.5, 0.01, effect_size_text, ha='center', fontsize=8, style='italic', color='#505050')

    
    return fig
```

```python
model_name_mapping = {
    "jinav3": "JINAv3",
    "gatortron_large": "Gatortron Large",
    "jinav3f": "JINAv3F",
    "nv_embed": "NV-embed-v2",
    "BioMistral-7B": "BioMistral-7B",
    "gpt-4o-2024-08-06": "GPT-4o"
}

key_metrics = ["Accuracy", "F1-macro", "AUROC"]
metric_keys = ["accuracy", "f1_macro", "aucroc"]

model_colors = {
    "JINAv3": '#808080',
    "Gatortron Large": '#70AD47',
    "BioMistral-7B": '#FFC000',
    "NV-embed-v2": '#4472C4',
    "JINAv3F": '#C00000',
    "GPT-4o": '#BA68C8' 
}

def extract_task_data(task_name, task_data):
    task_values = {}
    task_errors = {}
    task_annotations = {}
    
    for model_key, model_data in task_data["models"].items():
        if model_key not in model_name_mapping:
            continue
            
        display_name = model_name_mapping[model_key]
        
        values = []
        errors = []
        annotations = []
        
        for metric_key in metric_keys:
            if metric_key in model_data:
                values.append(model_data[metric_key]["value"])
                errors.append(model_data[metric_key]["std"])
                annotations.append(model_data[metric_key].get("effect", ""))
            else:
                values.append(0)
                errors.append(0)
                annotations.append("")
        
        task_values[display_name] = values
        task_errors[display_name] = errors
        task_annotations[display_name] = annotations
    
    return task_values, task_errors, task_annotations

tumor_type_data = extract_task_data("tumor_type", results["tumor_type"])
survival_data = extract_task_data("survival", results["survival"])
mgmt_data = extract_task_data("mgmt", results["mgmt"])

task1_fig = create_horizontal_task_plot(*tumor_type_data, "Tumor Type Classification")
task1_fig.savefig('tumor_type_classification.png', dpi=300, bbox_inches='tight')

task2_fig = create_horizontal_task_plot(*mgmt_data, "MGMT Methylation")
task2_fig.savefig('mgmt_methylation.png', dpi=300, bbox_inches='tight')

task3_fig = create_horizontal_task_plot(*survival_data, "One-Year Survival")
task3_fig.savefig('one_year_survival.png', dpi=300, bbox_inches='tight')
```
